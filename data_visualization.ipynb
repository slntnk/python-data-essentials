```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
import pandas_datareader as pdr
```

```python
#lineplot
```

```python
years=range(2010,2025)
```

```python
plt.plot(years)
```

```python
sales_x=[50,60,50,70,80,60,90,110,120,140,200,210,200,230,240]
```

```python
plt.plot(years,sales_x)
plt.xlabel("years")
plt.ylabel("Sales of product x (thousand)")
```

```python
sales_y=np.random.randint(100, 200, size=15)
sales_y
```

```python
plt.plot(years,sales_x)
plt.plot(years,sales_y)
plt.xlabel("years")
plt.ylabel("Sales of product x and y (thousand)")
```

```python
plt.plot(years,sales_x)
plt.plot(years,sales_y)
plt.xlabel("years")
plt.ylabel("Sales of product x and y (thousand)")
plt.title("Sales of product X and Y")
plt.legend(["X","Y"])
```

```python
plt.plot(years, sales_x, marker="*", c="purple", alpha=.5)
plt.plot(years, sales_y, marker="o", c="purple", alpha=1)
plt.xlabel("years")
plt.ylabel("Sales of product x and y (thousand)")
plt.title("Sales of product X and Y")
plt.legend(["X","Y"])
```

```python
plt.plot(years, sales_x, marker="*", c="green", ls="--", lw=2  )
plt.plot(years, sales_y, marker="o", c="red",ls ="-" , lw=4 )
plt.xlabel("years")
plt.ylabel("Sales of product x and y (thousand)")
plt.title("Sales of product X and Y")
plt.legend(["X","Y"])
```

```python
plt.figure(figsize=(8,4))
plt.plot(years, sales_x,)
plt.plot(years, sales_y,)
plt.xlabel("years")
plt.ylabel("Sales of product x and y (thousand)")
plt.title("Sales of product X and Y")
plt.legend(["X","Y"])
```
