```python
# Loading one of the built-in datasets from the Seaborn library.
import seaborn as sns
dfn=sns.load_dataset("diamonds")
```

```python
dfn.head()
```

```python
dfn.info()
```

```python
np.mean(dfn.select_dtypes(include=[np.number]), axis=0)
```

```python
#groupby
dfn.groupby("cut")
```

```python
dfn.groupby("cut")["price"].mean()
```

```python
dfn.groupby("color")["price"].mean()
```

```python
dfn.groupby("carat")["price"].describe()
```

```python
#aggregate
dfn.groupby("color")["price"].aggregate([ min, np.mean, np.median, max])
```

```python
dfn.groupby("color").aggregate({"carat": np.mean,"price": np.mean})
```

```python
#filter
dfn.head()
```

```python
dfn.filter(items=["carat","price"]).head(10)
```

```python
dfn_filtered = dfn.groupby("cut").filter(lambda x: (x["x"].mean() > dfn["x"].mean()) &
                                                   (x["y"].mean() > dfn["y"].mean()) &
                                                   (x["z"].mean() > dfn["z"].mean()))
result = dfn_filtered.groupby("cut")["price"].mean()
print(result)
```

```python
dfn_filtered2 = dfn.groupby("color").filter(lambda x: x["price"].mean() > 1000)

print(dfn_filtered2.groupby("color")["price"].mean())
```

```python
#transform
```

```python
dfn_mean=dfn.groupby("cut")["price"].transform("mean")
dfn_mean
```

```python
#apply
```

```python
dfn.groupby("cut")["price"].apply(np.mean)
```

```python
#pivot table
```

```python
dfn.groupby(["cut","color"])[["price"]].aggregate("mean").unstack()
# If double square brackets are added, it transforms into a DataFrame form.
# unstack(): "It is a function that converts one of the row indexes of a multi-level DataFrame into columns.
```

```python
dfn.pivot_table("price", index="cut", columns="color")
# A simpler version of the previous code using a pivot table."
```
